Resumen del notebook de limpieza

El objetivo de este notebook fue partir de la base original obtenida mediante scraping y construir una base limpia y consistente de departamentos usados en venta en CABA, orientada al posterior análisis de valuación, detección de propiedades subvaluadas y oportunidades de flipping.

La base original contenía 16.993 registros y 29 variables. El criterio de trabajo fue principalmente conservador: no eliminar observaciones simplemente por tener valores extremos o datos poco frecuentes, sino revisar si existía una razón concreta para considerar que una propiedad estaba fuera del universo de análisis o tenía información incorrecta.

1. Definición del universo de análisis

Primero se filtraron únicamente las categorías correspondientes a departamentos, excluyendo casas, PH originalmente clasificados como tales, terrenos, locales, oficinas, cocheras y otras tipologías.

También se verificó que las propiedades fueran operaciones de venta y que los precios estuvieran expresados en USD.

Como el objetivo es analizar departamentos usados, se buscaron y eliminaron publicaciones correspondientes a obra nueva, pozo, preventa, emprendimientos, desarrollos y propiedades a estrenar. Para esto no se utilizó únicamente la variable Antiguedad, ya que contenía valores negativos y ceros que podían tener distintas interpretaciones. Se combinaron esos datos con información extraída del título y la descripción.

2. Control de calidad de las variables

Se revisaron los valores faltantes, tipos de datos, duplicados y valores potencialmente inconsistentes.

Los duplicados no se eliminaron automáticamente. Por ejemplo, se encontraron publicaciones que compartían el mismo Link, pero correspondían a unidades diferentes dentro de un mismo desarrollo. Por eso se utilizaron principalmente ID e ID_Entidad para identificar las propiedades.

También se controlaron:

precios iguales o menores a cero;
superficies totales y cubiertas;
superficies cubiertas mayores a la superficie total;
cantidad de ambientes, dormitorios y baños;
antigüedad;
precio por metro cuadrado;
expensas y su moneda;
coordenadas geográficas;
barrios y ubicaciones faltantes.

Cuando se detectaron errores claros, se corrigieron sin eliminar innecesariamente la propiedad. Por ejemplo, una superficie cubierta igual a cero se transformó en faltante, un monoambiente con Ambientes = 0 se corrigió a 1 y algunas expensas o monedas evidentemente mal cargadas fueron corregidas o convertidas en faltantes.

3. Duplicados y propiedades no comparables

Se buscaron duplicados utilizando distintas combinaciones de variables. Se eliminaron únicamente aquellos registros que representaban realmente la misma propiedad repetida.

También se hizo una revisión adicional de propiedades que, aunque estaban clasificadas como departamentos en la fuente, podían corresponder a PH u otras situaciones especiales. Esto permitió eliminar 60 propiedades consideradas PH, manteniendo algunos casos donde el texto aclaraba que seguían siendo departamentos aunque tuvieran características similares a un PH.

Durante la revisión de valores extremos también se detectaron casos que no eran comparables con una compraventa residencial convencional. Por ejemplo, se eliminó una nuda propiedad con usufructo vitalicio, cuyo precio reducido respondía a una modalidad jurídica diferente, y una propiedad originalmente destinada a oficina profesional, aunque permitiera su transformación posterior en vivienda.

En cambio, no se eliminaron outliers solamente por presentar un precio/m² muy alto o muy bajo. Se revisaron los casos extremos y se comprobó que muchos podían explicarse por el barrio, el estado de conservación, grandes superficies descubiertas, cocheras o características premium.

4. Uso de título y descripción

Una parte importante del trabajo fue aprovechar las variables textuales Titulo y Descripcion para obtener información que no estaba estructurada originalmente.

Se creó la variable Estado_Propiedad, clasificando los departamentos en:

Necesita_obra
Buen_estado
Reciclada_refaccionada
Sin_clasificar

Para hacerlo se utilizaron expresiones presentes en los avisos, como “a refaccionar”, “a reciclar”, “requiere actualización”, “buen estado”, “excelente estado”, “reciclado”, “refaccionado” o “completamente remodelado”.

Se optó por mantener Sin_clasificar cuando el texto no daba evidencia suficiente, en lugar de asumir artificialmente el estado de la propiedad. Finalmente quedaron 416 departamentos que necesitan obra, 2.000 en buen estado, 1.085 reciclados/refaccionados y 6.170 sin clasificación.

Esta variable será especialmente importante para analizar la hipótesis relacionada con el potencial de flipping. Como primer control descriptivo, ya se observó que los departamentos que necesitan obra presentan un precio/m² considerablemente menor que aquellos en buen estado o reciclados, aunque esta relación deberá analizarse posteriormente controlando otras características.

5. Creación de variables de amenities

La variable Amenities contenía múltiples características separadas dentro de un mismo campo. En lugar de generar una variable para cada uno de los aproximadamente 95 valores encontrados, se seleccionaron aquellos con mayor utilidad para el análisis inmobiliario.

Se crearon variables binarias como:

Tiene_Balcon, Tiene_Ascensor, Tiene_Terraza, Tiene_Parrilla, Tiene_SUM, Tiene_Seguridad, Tiene_Aire_Acondicionado, Tiene_Baulera, Tiene_Patio, Tiene_Laundry, Tiene_Gimnasio, Tiene_Solarium y Tiene_Pileta.

También se creó Tiene_Cochera utilizando título y descripción. En este caso fue necesario distinguir entre una cochera realmente incluida en la venta y frases como “cochera opcional”, “posibilidad de comprar cochera” o “cochera no incluida”.

6. Coordenadas y ubicación

Se revisaron las coordenadas geográficas para detectar propiedades ubicadas aparentemente fuera de CABA. Se encontraron cinco casos donde la dirección y el barrio correspondían a CABA, pero las coordenadas eran claramente incorrectas.

Como la propiedad en sí seguía siendo válida, no se eliminaron esos registros: solamente se reemplazaron las coordenadas erróneas por valores faltantes.

Además, se decidió conservar el Barrio informado por la fuente, incluso cuando utiliza zonas comerciales como Palermo Chico o Barrio Norte. Más adelante se podrá realizar un cruce espacial utilizando latitud y longitud para obtener una clasificación oficial de barrios/comunas.
